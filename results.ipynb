```python
s = 10
A = np.random.randint(-4, 4, size=(s, s))
print('size:', A.shape)
print(A)
S = np.dot(A.T, A)
```

```python
print('Test if A @ A.T is symetric:')
print(is_sym(S))
```

```python
print('For matrix', A.shape)
start = timeit.default_timer()
detA = determinant_naive(A)
stop = timeit.default_timer()
print('determinant of A (naive method):', detA)
print(f'Computed in {stop - start:.2f} seconds')


start = timeit.default_timer()
detA = determinant_gauss(A)
stop = timeit.default_timer()
print('determinant of A (gaussian method):', detA)
print(f'Computed in {stop - start:.2f} seconds')


start = timeit.default_timer()
detA = np.linalg.det(A)
stop = timeit.default_timer()
print('determinant of A (numpy linalg mehtod):', detA)
print(f'Computed in {stop - start:.2f} seconds')
```

```python
print('Test if A @ A.T is positive definite:')
print(is_def_pos_sylvester(S))
```

```python
print('Test the inverse of A with gaussian with A @ inv(A) == Id:')
invA = inverse_gauss(A)
print(np.round(np.dot(invA, A) + 0.0001, 3))
```

```python
print('Test Cholesky algorithme on S = A @ A.T:')
L = cholesky(S)
if np.linalg.norm(np.dot(L, L.T) - S) < 0.001:
    print(True)
else:
    print(False)
```

```python
s = 4
A = np.random.randint(-4, 4,size=(s, s))
print('size:', A.shape)
print(A)
```

```python
print('Test the diagonalization algorithm with power iteration method')
print('For matrix', A.shape)
A_eigvals, A_eigvect = eigen(A)
while len(A_eigvals) != s:
    A = np.random.randint(-4, 4, size=(s, s))
    A_eigvals, A_eigvect = eigen(A)
print('Eigen value:', A_eigvals)
```

```python
s = 500
A = np.random.rand(s, s) / 10
S = np.dot(A.T, A)
print('size:', S.shape)
print('Test the diagonalization algorithm with power iteration method')
print('For matrix', S.shape)

start = timeit.default_timer()
S_eigvals, S_eigvect = eigen(S)
stop = timeit.default_timer()
print(f'Computed in {stop - start:.2f} seconds')
```

```python
print('eigen values:\n', S_eigvals)
S_eigvals2 = np.linalg.eigvals(S)
print('eigen values with numpy linalg:\n', S_eigvals2)
if sum(S_eigvals) - S_eigvals2.sum() < 1:
    print('True')
else:
    print('False')
```
